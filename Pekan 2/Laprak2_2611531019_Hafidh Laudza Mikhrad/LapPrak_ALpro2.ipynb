## 2.1 Dasar Teori

### 1. Pengenalan Bahasa Pemrograman Python

Python merupakan bahasa pemrograman tingkat tinggi (*high-level programming language*) yang bersifat *open-source*, terinterpretasi, dan berorientasi pada objek (*object-oriented*). Python dikenal dengan sintaksisnya yang bersih dan ringkas sehingga mengutamakan keterbacaan kode (*code readability*). Dalam praktiknya, Python mengadopsi mekanisme *dynamic typing*, yang berarti jenis tipe data pada suatu variabel didefinisikan secara otomatis oleh interpreter saat program dijalankan (*runtime*) sesuai dengan nilai yang diisikan ke dalamnya.

### 2. Konsep Variabel dan Tipe Data

Dalam pembuatan program, komputer membutuhkan lokasi memori untuk menyimpan informasi sementara:

* **Variabel** merupakan nama pengenal (*identifier*) atau wadah simbolis yang mengacu pada alokasi memori tertentu tempat data disimpan.
* **Tipe Data** adalah klasifikasi yang memberi tahu komputer mengenai jenis nilai yang dikandung oleh suatu variabel, sehingga komputer mengetahui jenis operasi matematika maupun logika apa yang dapat diterapkan pada data tersebut.

---

### 3. Tipe Data Skalar

Tipe data skalar (*primitive/scalar types*) adalah jenis tipe data paling dasar yang hanya mampu menampung satu nilai tunggal pada satu waktu.

#### A. Tipe Data Numerik (*Numeric Types*)

Tipe data numerik digunakan untuk menangani angka dan perhitungan matematis. Dalam Python, kelompok numerik terbagi menjadi tiga jenis:

1. **Integer (`int`)**: Menyimpan bilangan bulat positif maupun negatif tanpa koma atau desimal (contoh: `-10`, `0`, `100`).
2. **Float (`float`)**: Menyimpan bilangan riil yang memiliki titik desimal (*floating-point*) atau dituliskannya menggunakan notasi eksponensial (contoh: `3.14`, `-0.05`).
3. **Complex (`complex`)**: Menyimpan bilangan kompleks yang terdiri dari bagian riil dan bagian imajiner yang ditandai dengan sufiks `j` (contoh: `2 + 3j`).

#### B. Tipe Data Skalar Lainnya

1. **String (`str`)**: Kumpulan karakter atau teks yang diapit oleh tanda petik tunggal (`'`) atau ganda (`"`) (contoh: `"Praktikum Python"`).
2. **Boolean (`bool`)**: Tipe data logika yang hanya memiliki dua kemungkinan nilai, yaitu `True` (benar) atau `False` (salah).

---

### 4. Pengambilan Masukan Data (*Input Data*)

Dalam aplikasi interaktif, program membutuhkan masukan dari pengguna saat runtime:

* **Fungsi `input()`**: Digunakan untuk menangkap masukan teks yang diketikkan pengguna melalui konsol/terminal.
* **Sifat Bawaan**: Nilai yang dikembalikan oleh fungsi `input()` secara bawaan akan selalu bertipe data **String (`str`)**.
* **Konversi Tipe Data (*Type Casting*)**: Apabila data masukan pengguna hendak diolah dalam perhitungan matematis, data string tersebut harus dikonversi terlebih dahulu ke tipe numerik yang sesuai menggunakan fungsi penataan tipe seperti `int()` atau `float()`.

---

### 5. Konsep Konstanta (*Constants*)

**Konstanta** adalah jenis variabel khusus yang nilainya bersifat tetap dan idealnya tidak boleh diubah selama eksekusi program berjalan.

* Berbeda dengan bahasa pemrograman lain seperti Java atau C++, Python tidak menyediakan kata kunci (*keyword*) khusus untuk mengunci nilai variabel (seperti `const` atau `final`).
* Sebagai gantinya, komunitas Python menyepakati aturan konvensi standar (PEP 8) di mana penulisan nama variabel konstanta menggunakan **HURUF KAPITAL ENTIRELY** (contoh: `PI = 3.14159`, `GRAVITASI = 9.8`). Hal ini berfungsi sebagai penanda visual bagi pemrogram agar tidak mengubah isi variabel tersebut.

### 2.2.2 Program numerik "A_2611531019"

```python
a_1019=100
print("The type of variable having value", a_1019, "is", type(a_1019))

b_1019=10.2345
print("The type of variable having value", b_1019, "is", type(b_1019))

c_1019=100+3j
print("The type of variable having value", c_1019, "is", type(c_1019))
```

**Penjelasan :**
1. Pada baris pertama dan kedua terdapat variable `a_1019` bernilai `100` yang merupakan bilangan bulat, sehingga fungsi `type()` mengembalikan tipe data `<class 'int'>`
2. Pada baris keempat dan kelima terdapat variabel `b_1019` bernilai `10.2345` yang merupakan bilangan desimal, sehingga fungsi `type()` mengembalikan tipe data `<class 'float'>`
3. Pada baris ketujuh dan kedelapan terdapat variable `c_1019` bernilai `100+3j` yang merupakan bilangan kompleks (memiliki bagian riil dan imajiner `j`), sehingga fungsi `type()` mengembalikan tipe data `<class 'complex'>`

### 2.2.3 Program konstanta "A_2611531019"

```python
from typing import Final
PI: Final = 3.14
print("pi: %f" % (PI))
jari_1019 = float(input("Masukkan jari-jari: "))
luas_1019 = PI * jari_1019 * jari_1019
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_1019, luas_1019))
```

**Penjelasan :**
1. Baris pertama mengimpor `Final` dari modul `typing` untuk menandai suatu variable sebagai konstanta agar nilainya tidak diubah
2. Baris kedua mendeklarasikan konstanta `PI` dengan nilai `3.14` menggunakan tipe hint `Final`
3. Baris ketiga menampilkan nilai konstanta `PI` ke layar menggunakan format string `%f`
4. Baris keempat mengambil input jari-jari dari pengguna, lalu mengonversinya menjadi tipe data desimal (`float`) dan menyimpannya dalam variable `jari_1019`
5. Baris kelima menghitung luas linkaran dengan rumus PI * jari * jari dan menyimpannya ke dalam variable `luas_1019`
6. Baris keenam menampilkan hasil perhitungan luas lingkaran dengan format 2 angka di belakang koma (`%.2f`)
