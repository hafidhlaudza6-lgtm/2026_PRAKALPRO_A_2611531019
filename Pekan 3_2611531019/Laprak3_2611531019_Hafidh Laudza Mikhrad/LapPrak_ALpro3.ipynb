## 2.1 Dasar Teori

Dalam praktikum ini dipelajari berbagai operator yang terdapat pada bahasa pemrograman Python. Operator yang dibahas meliputi operator aritmatika, assignment, bitwise, logika, keanggotaan, dan perbandingan. Setiap operator memiliki fungsi yang berbeda dalam proses pengolahan data. Penggunaan operator yang tepat dapat membantu program melakukan perhitungan, memproses kondisi, serta menghasilkan keluaran sesuai dengan tujuan yang telah ditentukan. Adapun penjelasan mengenai operator yang dipelajari adalah sebagai berikut:

## 2.1.1 Operator Aritmatika

Operator aritmatika digunakan untuk melakukan operasi matematika pada nilai numerik. Beberapa operator yang termasuk di dalamnya adalah penjumlahan (+), pengurangan (-), perkalian (*), pembagian (/), pembagian bulat (//), modulus (%), dan perpangkatan (**). Operator pembagian bulat digunakan untuk mendapatkan hasil bagi dalam bentuk bilangan bulat, sedangkan modulus digunakan untuk memperoleh sisa hasil pembagian. Jenis operator ini umumnya digunakan dalam program yang membutuhkan proses penghitungan atau manipulasi nilai angka.

## 2.1.2 Operator Assignment

Assignment merupakan operator yang berfungsi untuk menetapkan nilai ke dalam sebuah variabel maupun melakukan perubahan terhadap nilai tersebut. Operator dasar yang digunakan adalah tanda sama dengan (=). Selain operator tersebut, Python menyediakan beberapa bentuk assignment gabungan, seperti +=, -=, *=, /=, //=, %=, dan **=. Dengan menggunakan assignment gabungan, nilai sebuah variabel dapat langsung diperbarui berdasarkan nilai yang telah dimilikinya. Hal tersebut membuat kode program menjadi lebih singkat dan praktis.

## 2.1.3 Operator Bitwise

Operator bitwise digunakan untuk memproses bilangan berdasarkan susunan bit atau bentuk binernya. Operator yang termasuk dalam kategori ini antara lain AND (&), OR (|), XOR (^), NOT (~), shift kiri (<<), dan shift kanan (>>). Operator AND, OR, serta XOR bekerja dengan membandingkan bit dari dua nilai. Operator NOT berfungsi membalik nilai setiap bit, sedangkan shift kiri dan shift kanan digunakan untuk menggeser posisi bit ke arah tertentu. Penggunaan operator bitwise biasanya ditemukan pada proses pemrograman yang membutuhkan manipulasi data secara langsung pada tingkat biner.

## 2.1.4 Operator Logika

Operator logika digunakan untuk melakukan pengolahan terhadap nilai boolean serta menggabungkan beberapa kondisi dalam sebuah ekspresi. Python memiliki tiga operator logika utama, yaitu and, or, dan not. Operator and menghasilkan True apabila seluruh kondisi yang diperiksa bernilai True. Operator or akan menghasilkan True apabila terdapat setidaknya satu kondisi yang bernilai True. Sementara itu, operator not berfungsi menghasilkan nilai kebalikan dari kondisi yang diberikan. Operator ini banyak digunakan dalam pembuatan program yang melibatkan beberapa persyaratan atau kondisi.

## 2.1.5 Operator Keanggotaan

Operator keanggotaan digunakan untuk mengetahui apakah suatu nilai terdapat dalam sebuah kumpulan data, misalnya list atau struktur data lainnya. Python menyediakan operator in dan not in. Operator in akan menghasilkan True jika nilai yang diperiksa terdapat dalam kumpulan data, sedangkan not in menghasilkan True jika nilai tersebut tidak ditemukan.

Selain operator keanggotaan, terdapat pula operator identitas yang digunakan untuk mengetahui apakah dua variabel mengacu pada objek yang sama. Operator identitas terdiri dari is dan is not. Berbeda dengan operator perbandingan (==) yang memeriksa kesamaan nilai, operator identitas memeriksa apakah kedua variabel menunjuk pada objek yang sama. Oleh karena itu, dua variabel dapat memiliki nilai yang sama tetapi belum tentu merujuk pada objek yang sama.

## 2.1.6 Operator Perbandingan

Operator perbandingan berfungsi untuk menentukan hubungan antara dua nilai. Hasil dari proses perbandingan berupa nilai boolean, yaitu True atau False. Operator yang digunakan meliputi lebih besar dari (>), lebih kecil dari (<), lebih besar atau sama dengan (>=), lebih kecil atau sama dengan (<=), sama dengan (==), serta tidak sama dengan (!=).

Python juga memungkinkan penggunaan perbandingan secara berantai dalam satu ekspresi. Contohnya adalah 0 < x < 100. Ekspresi tersebut digunakan untuk memeriksa apakah nilai x berada di antara 0 dan 100 dengan melakukan dua perbandingan secara berurutan.

### 2.2.1 Program aritmatika "A_2611531019"

````python
# buat file dengan nama aritmatika_NIM.py
# buat program untuk operator aritmatika dalam Python
# Nama variable ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer

angka1_1019 = int(input("Input angka-1: "))
angka2_1019 = int(input("Input angka-2: "))

# penjumlahan
hasil = angka1_1019 + angka2_1019
print("\nOperator Penjumlahan")
print("Hasil =", hasil)

# pengurangan
hasil = angka1_1019 - angka2_1019
print("\nOperator Pengurangan")
print("Hasil =", hasil)

# perkalian
hasil = angka1_1019 * angka2_1019
print("\nOperator Perkalian")
print("Hasil =", hasil)

# pembagian, pembagian bulat, dan sisa bagi
if angka2_1019 != 0:
    hasil = angka1_1019 / angka2_1019
    print("\nOperator Pembagian")
    print("Hasil =", hasil)

    hasil = angka1_1019 // angka2_1019
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil)

    hasil = angka1_1019 % angka2_1019
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil)
else:
    print("Angka kedua tidak boleh bernilai 0")

# pangkat
hasil = angka1_1019 ** angka2_1019
print("\nOperator Pangkat")
print("Hasil =", hasil)
````

### 2.2.2 Program assignment "A_2611531019"

````python
# Buat file dengan nama assignment_NIM.py
# Nama variable ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator assignment dalam Python

angka1_1019 = int(input("Input angka-1: "))
angka2_1019 = int(input("Input angka-2: "))

print("\nNilai awal angka1_1019 =", angka1_1019)
print("Nilai angka2_1019 =", angka2_1019)

# Assignment biasa
hasil = angka1_1019
print("\nAssignment biasa (=)")
print("Hasil =", hasil)

# Assignment penambahan
hasil = angka1_1019
hasil += angka2_1019
print("\nAssignment penambahan (+=)")
print("Hasil =", hasil)

# Assignment pengurangan
hasil = angka1_1019
hasil -= angka2_1019
print("\nAssignment pengurangan (-=)")
print("Hasil =", hasil)

# Assignment perkalian
hasil = angka1_1019
hasil *= angka2_1019
print("\nAssignment perkalian (*=)")
print("Hasil =", hasil)

# Assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_1019 != 0:
    hasil = angka1_1019
    hasil /= angka2_1019
    print("\nAssignment pembagian (/=)")
    print("Hasil =", hasil)
    # Operator tambahan
    hasil = angka1_1019
    hasil //= angka2_1019
    print("\nAssignment pembagian bulat (//=)")
    print("Hasil =", hasil)
    hasil = angka1_1019
    hasil %= angka2_1019
    print("\nAssignment sisa bagi (%=)")
    print("Hasil =", hasil)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0")

# Operator tambahan: assignment perpangkatan
hasil = angka1_1019
hasil **= angka2_1019
print("\nAssignment perpangkatan (**=)")
print("Hasil =", hasil)
````

### 2.2.3 Program bitwise "A_2611531019"

````python
# Buat file dengan nama bitwise_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()

print("\n===================================")
print("3. OPERATOR BITWISE")
print("===================================")

angka1_1019 = int(input("Masukkan angka bitwise-1: "))
angka2_1019 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1_1019 =", angka1_1019, "| biner =", bin(angka1_1019))
print("angka2_1019 =", angka2_1019, "| biner =", bin(angka2_1019))

# Bitwise AND
hasil = angka1_1019 & angka2_1019
print("\nBitwise AND (&)")
print(angka1_1019, "&", angka2_1019, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, '08b'))

# Bitwise OR
hasil = angka1_1019 | angka2_1019
print("\nBitwise OR (|)")
print(angka1_1019, "|", angka2_1019, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, '08b'))

# Bitwise XOR
hasil = angka1_1019 ^ angka2_1019
print("\nBitwise XOR (^)")
print(angka1_1019, "^", angka2_1019, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, '08b'))

# Bitwise NOT
hasil = ~angka1_1019
print("\nBitwise NOT (~)")
print("~", angka1_1019, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, '08b'))

# Bitwise geser kiri
jumlah_geser = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil = angka1_1019 << jumlah_geser
print("\nBitwise Geser Kiri (<<)")
print(angka1_1019, "<<", jumlah_geser, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, '08b'))

# Bitwise geser kanan
hasil = angka1_1019 >> jumlah_geser
print("\nBitwise Geser Kanan (>>)")
print(angka1_1019, ">>", jumlah_geser, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, '08b'))
````


* **Fungsi**:
* Menampilkan nilai desimal asli dari kedua variabel.
* `bin(...)`: Berfungsi untuk mengubah nilai desimal menjadi representasi string biner (diawali dengan prefiks `0b`).



---

#### **4. Bagian Operasi Bitwise AND (`&`)**

```python
hasil = angka1_1019 & angka2_1019
print("\nBitwise AND (&)")
print(angka1_1019, "&", angka2_1019, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, '08b'))

```

* **Fungsi**:
* `angka1_1019 & angka2_1019`: Membandingkan bit satu per satu. Bit bernilai `1` hanya jika **kedua bit yang dibandingkan bernilai 1**.
* `format(hasil, '08b')`: Menampilkan hasil biner agar konsisten sepanjang **8-bit** dengan penambahan angka `0` di depan jika diperlukan.



---

#### **5. Bagian Operasi Bitwise OR (`|`)**

```python
hasil = angka1_1019 | angka2_1019
print("\nBitwise OR (|)")
print(angka1_1019, "|", angka2_1019, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, '08b'))

```

* **Fungsi**:
* `angka1_1019 | angka2_1019`: Menghasilkan bit `1` jika **salah satu atau kedua bit yang dibandingkan bernilai 1**.



---

#### **6. Bagian Operasi Bitwise XOR (`^`)**

```python
hasil = angka1_1019 ^ angka2_1019
print("\nBitwise XOR (^)")
print(angka1_1019, "^", angka2_1019, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, '08b'))

```

* **Fungsi**:
* `angka1_1019 ^ angka2_1019`: (Exclusive OR). Menghasilkan bit `1` jika **kedua bit yang dibandingkan berbeda** (satu `0` dan satu `1`), dan `0` jika sama.



---

#### **7. Bagian Operasi Bitwise NOT (`~`)**

```python
hasil = ~angka1_1019
print("\nBitwise NOT (~)")
print("~", angka1_1019, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, '08b'))

```

* **Fungsi**:
* `~angka1_1019`: Membalikkan seluruh nilai bit (operasi *bitwise complement*), di mana angka `0` menjadi `1` dan sebaliknya, mengikuti perhitungan matematis komputer $-(x + 1)$.



---

#### **8. Bagian Operasi Pergeseran Bit (Shift)**

```python
jumlah_geser = int(input("\nMasukkan jumlah pergeseran bit: "))
```

* **Fungsi**: Meminta input dari pengguna mengenai seberapa banyak posisi bit yang ingin digeser.

```python
hasil = angka1_1019 << jumlah_geser

```

* **Fungsi**: `<<` Menggeser posisi bit ke kiri sebanyak nilai `jumlah_geser` (setara dengan perkalian matematika dengan $2$ pangkat `jumlah_geser`).

```python
hasil = angka1_1019 >> jumlah_geser

```

* **Fungsi**: `>>` Menggeser posisi bit ke kanan sebanyak nilai `jumlah_geser` (setara dengan pembagian bulat dengan $2$ pangkat `jumlah_geser`).

### 2.2.4 Program lainnya "A_2611531019"

````python
# Buat file dengan nama lainnya_NIM.py
# Nama variable ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Program operator keanggotaan dan identitas

print("==================================")
print("1. OPERATOR KEANGGOTAAN")
print("==================================")

# Input beberapa data yang dipisahkan dengan koma
input_data = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# Mengubah input menjadi list integer
data = [int(angka.strip()) for angka in input_data.split(",")]

nilai_dicari = int(input("Masukkan nilai yang ingin dicari: "))

# Operator in
hasil_1019 = nilai_dicari in data
print("\nOperator keanggotaan IN")
print(nilai_dicari, "in", data,"=", hasil_1019)

# Operator not in
hasil_1019 = nilai_dicari not in data
print("\nOperator keanggotaan NOT IN")
print(nilai_dicari, "not in", data,"=", hasil_1019)

print("\n==================================")
print("2. OPERATOR IDENTITAS")
print("==================================")

# objek1 menggunakan list dari input pengguna
objek1_1019 = data

# objek2 merujuk pada objek yang sama dengan objek1
objek2_1019 = objek1_1019

# objek3 memiliki isi sama, tetapi merupakan objek baru
objek3_1019 = data.copy()

print("objek1_1019 =", objek1_1019)
print("objek2_1019 =", objek2_1019)
print("objek3_1019 =", objek3_1019)

# Operator is
hasil_1019 = objek1_1019 is objek2_1019
print("\nOperator identitas IS")
print("objek1_1019 is objek2_1019 =", hasil_1019)

# Operator is not
hasil_1019 = objek1_1019 is not objek3_1019
print("\nOperator identitas IS NOT")
print("objek1_1019 is not objek3_1019 =", hasil_1019)

# Membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai")
print("objek1_1019 is objek3_1019:", objek1_1019 is objek3_1019)
print("objek1_1019 == objek3_1019:", objek1_1019 == objek3_1019)
````

### **Pejelasan**

---

#### **Bagian 1: Operator Keanggotaan (Membership Operators)**

Operator keanggotaan digunakan untuk menguji apakah suatu nilai atau elemen tertentu terdapat di dalam sebuah urutan (seperti *list*, *tuple*, atau *string*).

```python
input_data = input("Masukkan beberapa angka, pisahkan dengan koma: ")
data = [int(angka.strip()) for angka in input_data.split(",")]

```

* **`input_data = ...`**: Menerima input dari pengguna dalam bentuk teks (*string*), misalnya memasukkan: `10, 20, 30`.
* **`input_data.split(",")`**: Memecah string input berdasarkan tanda koma `,` menjadi bentuk list string (contoh: `['10', ' 20', ' 30']`).
* **`angka.strip()`**: Menghilangkan spasi ekstra yang mungkin tertinggal di sekitar angka setelah dipisah.
* **`int(...)` & *List Comprehension***: Mengubah setiap elemen string yang sudah bersih tadi menjadi tipe data **integer** secara otomatis ke dalam variabel `data`.

```python
nilai_dicari = int(input("Masukkan nilai yang ingin dicari: "))

```

* Meminta pengguna memasukkan satu angka spesifik yang ingin dicek keberadaannya, lalu dikonversi menjadi integer.

```python
# Operator in
hasil_1019 = nilai_dicari in data
print(nilai_dicari, "in", data, "=", hasil_1019)

# Operator not in
hasil_1019 = nilai_dicari not in data
print(nilai_dicari, "not in", data, "=", hasil_1019)

```

* **`in`**: Menghasilkan nilai `True` jika `nilai_dicari` ditemukan di dalam list `data`, dan `False` jika tidak ada.
* **`not in`**: Menghasilkan nilai `True` jika `nilai_dicari` **tidak** ditemukan di dalam list `data`.

---

#### **Bagian 2: Operator Identitas (Identity Operators)**

Operator identitas digunakan untuk membandingkan lokasi memori dari dua objek, bukan sekadar membandingkan nilai isinya.

```python
objek1_1019 = data
objek2_1019 = objek1_1019
objek3_1019 = data.copy()

```

* **`objek1_1019 = data`**: Variabel `objek1_1019` merujuk ke list `data`.
* **`objek2_1019 = objek1_1019`**: Variabel `objek2_1019` menunjuk ke objek memori yang **sama persis** dengan `objek1_1019`.
* **`objek3_1019 = data.copy()`**: Membuat sebuah **salinan baru** dari list `data`. Meskipun isi nilainya sama persis dengan `objek1_1019`, ia menempati alokasi memori yang berbeda.

```python
# Operator is
hasil_1019 = objek1_1019 is objek2_1019
print("objek1_1019 is objek2_1019 =", hasil_1019)

# Operator is not
hasil_1019 = objek1_1019 is not objek3_1019
print("objek1_1019 is not objek3_1019 =", hasil_1019)

```

* **`is`**: Menghasilkan `True` jika kedua variabel menunjuk ke **objek memori yang sama** (berlaku pada `objek1_1019` dan `objek2_1019`).
* **`is not`**: Menghasilkan `True` jika kedua variabel menunjuk ke **lokasi memori yang berbeda** (berlaku pada `objek1_1019` dan `objek3_1019` karena menggunakan `.copy()`).


### 2.2.5 Program logika "A_2611531019"

````python
# buat nama file dengan nama logika_NIM.py
# nama variable ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# program operator logika dalam Python

# Masukkan nilai boolean
# Input tidak peka terhadap huruf besar dan kecil
a1 = input("Input nilai boolean-1 (True/False): ").strip().lower() == "true"
a2 = input("Input nilai boolean-2 (True/False): ").strip().lower() == "true"

print("\nA1 =", a1)
print("A2 =", a2)

# konjungsi: bernilai True jika keduanya True
hasil = a1 and a2
print("\nKonjungsi (AND)")
print("A1 and A2 =", hasil)

# Disjungsi: bernilai True jika salah satunya True
hasil = a1 or a2
print("\nDisjungsi (OR)")
print("A1 or A2 =", hasil)

# Negasi A1: membalik nilai A1
hasil = not a1
print("\nNegasi A1 (NOT)")
print("not A1 =", hasil)

# Negasi A2: membalik nilai A2
hasil = not a2
print("\nNegasi A2 (NOT)")
print("not A2 =", hasil)

# XOR: bernilai true jika kedua nilai berbeda
hasil = a1 != a2
print("\nDisjungsi Eksklusif (XOR)")
print("A1 XOR A2 =", hasil)
````

Jika ya, variabel `a1` atau `a2` otomatis bernilai boolean `True`; jika bukan, bernilai `False`. *(Catatan: Sesuai aturan penamaan variabel dengan 4 digit NIM, nama variabel ini sebenarnya bisa disesuaikan menjadi `a1_1019` dan `a2_1019`).*

```python
hasil = a1 and a2
```

* **Fungsi**:

### 2.2.6 Program perbandingan "A_2611531019"

````python
# buat nama file dengan nama perbandingan_NIM.py
# nama variable ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# program operator perbandingan dalam Python

angka1_1019 = int(input("Input angka-1: "))
angka2_1019 = int(input("Input angka-2: "))

# Lebih besar dari
hasil = angka1_1019 > angka2_1019
print("/nOperator lebih besar dari")
print("angka1_1019 > angka2_1019 =", hasil)

# Lebih kecil dari
hasil = angka1_1019 < angka2_1019
print("/nOperator lebih kecil dari")
print("angka1_1019 < angka2_1019 =", hasil)

# Lebih besar dari atau sama dengan
hasil = angka1_1019 >= angka2_1019
print("/nOperator lebih besar dari atau sama dengan")
print("angka1_1019 >= angka2_1019 =", hasil)

# Lebih kecil dari atau sama dengan
hasil = angka1_1019 <= angka2_1019
print("/nOperator lebih kecil dari atau sama dengan")
print("angka1_1019 <= angka2_1019 =", hasil)

# Sama dengan
hasil = angka1_1019 == angka2_1019
print("/nOperator sama dengan")
print("angka1_1019 == angka2_1019 =", hasil)

# Tidak sama dengan
hasil = angka1_1019 != angka2_1019
print("/nOperator tidak sama dengan")
print("angka1_1019 != angka2_1019 =", hasil)

# Tambahan: perbandingan berantai dalam Python
hasil = 0 < angka1_1019 < 100
print("/nPerbandingan berantai")
print("0 < angka1_1019 < 100 =", hasil)

hasil = 0 < angka2_1019 < 100
print("0 < angka2_1019 < 100 =", hasil)
````
