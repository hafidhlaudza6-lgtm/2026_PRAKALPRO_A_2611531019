## 2.1 Dasar Teori

Struktur percabangan (Control Flow / Branching) adalah suatu instruksi dalam pemrograman yang memungkinkan alur eksekusi program berjalan secara dinamis berdasarkan suatu kondisi bernilai boolean (`True` atau `False`). Pada bahasa pemograman Python, terdapat beberapa jenis struktur percabangan yang digunakan untuk mengakomodasi berbagai tingkat kerumitan logika:

### 2.1.1 Pernyataan `if`

Digunakan untuk mengeksekusi satu blok kode tertentu hanya jika kondisi yang diuji bernilai `True`.

### 2.1.2 Pernyataan `if-else`

Menyediakan dua jalur eksekusi, di mana blok kode pertama dijalankan jika kondisi bernilai `True`, dan blok kode alternatif (`else`) dijalankan jika kondisi bernilai `False`.

### 2.1.3 Pernyataan `if-elif-else`

Digunakan ketika terdapat lebih dari dua alternatif kondisi yang harus diperiksa secara berurutan. Python akan mengevaluasi kondisi dari atas ke bawah dan berhenti pada kondisi pertama yang bernilai `True`.

### 2.1.4 Multi-if (If Berganda)

Merupakan sekumpulan pernyataan `if` terpisah yang dijalankan secara berurutan, sehingga setiap kondisi akan diuji terlepas dari hasil kondisi sebelumnya (memungkinkan beberapa blok kode tereksekusi sekaligus jika memenuhi syarat).

### 2.1.5 Pernyataan `match-case`

Diperkenalkan mulai Python 3.10, struktur ini berfungsi untuk mencocokkan nilai variabel terhadap sejumlah pola (`case`) secara lebih bersih dan efisien dibandingkan `if-elif-else` bertingkat.

### 2.2.4 Program lainnya "A_2611531019"

````python
# Buat file dengan nama lainnya_NIM.py
# Nama variable ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Program operator keanggotaan dan identitas

print("==================================")
print("1. OPERATOR KEANGGOTAAN")
print("==================================")

# Input beberapa data yang dipisahkan dengan koma
input_data = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# Mengubah input menjadi list integer
data = [int(angka.strip()) for angka in input_data.split(",")]

nilai_dicari = int(input("Masukkan nilai yang ingin dicari: "))

# Operator in
hasil_1019 = nilai_dicari in data
print("\nOperator keanggotaan IN")
print(nilai_dicari, "in", data,"=", hasil_1019)

# Operator not in
hasil_1019 = nilai_dicari not in data
print("\nOperator keanggotaan NOT IN")
print(nilai_dicari, "not in", data,"=", hasil_1019)

print("\n==================================")
print("2. OPERATOR IDENTITAS")
print("==================================")

# objek1 menggunakan list dari input pengguna
objek1_1019 = data

# objek2 merujuk pada objek yang sama dengan objek1
objek2_1019 = objek1_1019

# objek3 memiliki isi sama, tetapi merupakan objek baru
objek3_1019 = data.copy()

print("objek1_1019 =", objek1_1019)
print("objek2_1019 =", objek2_1019)
print("objek3_1019 =", objek3_1019)

# Operator is
hasil_1019 = objek1_1019 is objek2_1019
print("\nOperator identitas IS")
print("objek1_1019 is objek2_1019 =", hasil_1019)

# Operator is not
hasil_1019 = objek1_1019 is not objek3_1019
print("\nOperator identitas IS NOT")
print("objek1_1019 is not objek3_1019 =", hasil_1019)

# Membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai")
print("objek1_1019 is objek3_1019:", objek1_1019 is objek3_1019)
print("objek1_1019 == objek3_1019:", objek1_1019 == objek3_1019)
````

#### **Bagian 1: Operator Keanggotaan (Membership Operators)**

Operator keanggotaan digunakan untuk menguji apakah suatu nilai atau elemen tertentu terdapat di dalam sebuah urutan (seperti *list*, *tuple*, atau *string*).

```python
input_data = input("Masukkan beberapa angka, pisahkan dengan koma: ")
data = [int(angka.strip()) for angka in input_data.split(",")]

```

* **`input_data = ...`**: Menerima input dari pengguna dalam bentuk teks (*string*), misalnya memasukkan: `10, 20, 30`.
* **`input_data.split(",")`**: Memecah string input berdasarkan tanda koma `,` menjadi bentuk list string (contoh: `['10', ' 20', ' 30']`).
* **`angka.strip()`**: Menghilangkan spasi ekstra yang mungkin tertinggal di sekitar angka setelah dipisah.
* **`int(...)` & *List Comprehension***: Mengubah setiap elemen string yang sudah bersih tadi menjadi tipe data **integer** secara otomatis ke dalam variabel `data`.

```python
nilai_dicari = int(input("Masukkan nilai yang ingin dicari: "))

```

* Meminta pengguna memasukkan satu angka spesifik yang ingin dicek keberadaannya, lalu dikonversi menjadi integer.

```python
# Operator in

print(nilai_dicari, "in", data, "=", hasil_1019)

# Operator not in
hasil_1019 = nilai_dicari not in data
print(nilai_dicari, "not in", data, "=", hasil_1019)

```

* **`in`**: Menghasilkan nilai `True` jika `nilai_dicari` ditemukan di dalam list `data`, dan `False` jika tidak ada.
* **`not in`**: Menghasilkan nilai `True` jika `nilai_dicari` **tidak** ditemukan di dalam list `data`.

---

#### **Bagian 2: Operator Identitas (Identity Operators)**

Operator identitas digunakan untuk membandingkan lokasi memori dari dua objek, bukan sekadar membandingkan nilai isinya.

```python
objek1_1019 = data
objek2_1019 = objek1_1019
objek3_1019 = data.copy()

```

* **`objek1_1019 = data`**: Variabel `objek1_1019` merujuk ke list `data`.
* **`objek2_1019 = objek1_1019`**: Variabel `objek2_1019` menunjuk ke objek memori yang **sama persis** dengan `objek1_1019`.
* **`objek3_1019 = data.copy()`**: Membuat sebuah **salinan baru** dari list `data`. Meskipun isi nilainya sama persis dengan `objek1_1019`, ia menempati alokasi memori yang berbeda.

```python
# Operator is
hasil_1019 = objek1_1019 is objek2_1019
print("objek1_1019 is objek2_1019 =", hasil_1019)

# Operator is not
hasil_1019 = objek1_1019 is not objek3_1019
print("objek1_1019 is not objek3_1019 =", hasil_1019)

```

* **`is`**: Menghasilkan `True` jika kedua variabel menunjuk ke **objek memori yang sama** (berlaku pada `objek1_1019` dan `objek2_1019`).
* **`is not`**: Menghasilkan `True` jika kedua variabel menunjuk ke **lokasi memori yang berbeda** (berlaku pada `objek1_1019` dan `objek3_1019` karena menggunakan `.copy()`).

### **Jawaban**

#### 1. Operator Aritmatika dalam Transaksi

Operator aritmatika digunakan untuk melakukan operasi matematika dasar. Dalam aplikasi transaksi, operator ini berperan penting untuk menghitung total biaya, potongan, dan kembalian:

* **Perkalian (`*`)**: Digunakan untuk menghitung subtotal dari suatu barang (Harga Satuan $\times$ Jumlah Beli).
* **Pengurangan (`-`)**: Digunakan untuk mengurangkan nilai total belanja dengan nominal diskon atau potongan harga.
* **Penjumlahan (`+`)**: Digunakan untuk menggabungkan subtotal beberapa barang atau menambahkan pajak/biaya layanan ke total tagihan.
* **Pembagian (`/` atau `//`)**: Digunakan untuk membagi total tagihan (misalnya jika tagihan ingin dibagi rata) atau menghitung rata-rata belanja.
* **Modulo (`%`)**: Digunakan untuk mengecek sisa bagi, yang sering dipakai dalam menentukan aturan diskon kelipatan tertentu.

#### 2. Perbedaan Operator Logika (`and`, `or`, `not`)

Operator logika digunakan untuk menggabungkan beberapa kondisi atau ekspresi boolean:

* **`and`**: Menghasilkan nilai `True` **hanya jika semua** kondisi yang dihubungkan bernilai benar. Contoh: Pelanggan mendapat diskon jika `total_belanja > 100000 and status_member == True`.
* **`or`**: Menghasilkan nilai `True` jika **salah satu atau kedua** kondisi bernilai benar. Contoh: Diskon berlaku jika `hari == "Minggu" or punya_kupon == True`.
* **`not`**: Berfungsi untuk **membalikkan nilai kebenaran** (negasi). Jika kondisinya `True`, maka `not` akan mengubahnya menjadi `False`, begitu pula sebaliknya. Contoh: `not is_blocked` memastikan pengguna tidak sedang diblokir.

#### 3. Fungsi Operator Keanggotaan (`in` dan `not in`)

Operator keanggotaan memeriksa keberadaan suatu nilai di dalam objek wadah (*sequence* atau *collection*) seperti *list*, *tuple*, atau *string*:

* **`in`**: Digunakan untuk memeriksa apakah sebuah kode promo yang dimasukkan pengguna **ada di dalam** daftar kode promo yang valid (misalnya: `"HEMAT50" in daftar_promo`). Menghasilkan `True` jika ditemukan.
* **`not in`**: Digunakan untuk memastikan bahwa sebuah nilai **tidak ada di dalam** wadah tertentu. Contohnya untuk memeriksa apakah kode promo yang dimasukkan belum pernah digunakan atau tidak terdaftar di daftar hitam (*blacklist*).

---

#### 4. Perbedaan `is` dan `==` pada Python

* **`==` (Equality)**: Membandingkan **nilai** dari dua objek. Operator ini bernilai `True` jika isi atau nilai dari kedua variabel tersebut sama, meskipun berada di tempat memori yang berbeda.
* **`is` (Identity)**: Membandingkan **identitas memori** dari dua objek. Operator ini bernilai `True` hanya jika kedua variabel tersebut merujuk pada objek fisik yang *sama persis* di dalam memori komputer (menunjuk ke alamat memori yang identik).

---

#### 5. Penggunaan Operator Bitwise untuk Hak Akses Pelanggan

Operator bitwise beroperasi pada tingkat bit biner individual (seperti `&`, `|`, `^`, `~`). Dalam sistem hak akses (*access control*):

* Setiap hak akses direpresentasikan oleh bit tertentu (misal: bit ke-0 untuk akses Baca, bit ke-1 untuk akses Tulis).
* **Operator AND (`&`)** digunakan untuk menyaring (*filtering* atau *masking*) apakah pelanggan memiliki izin tertentu. Contoh: `hak_akses & IZIN_BACA` akan bernilai bukan nol jika pelanggan memiliki izin baca.
* **Operator OR (`|`)** digunakan untuk menggabungkan beberapa hak akses baru ke akun pelanggan.

---

#### 6. Alur Program dari Input hingga Pembayaran dan Status Promo

1. **Input Data**: Pengguna memasukkan data transaksi seperti nama barang, jumlah, harga, serta kode promo melalui terminal/antarmuka.
2. **Pemeriksaan Promo**: Program menggunakan operator `in` untuk mencocokkan kode promo dan operator logika untuk memvalidasi syarat minimum belanja.
3. **Penghitungan Aritmatika**:
* Program menghitung subtotal menggunakan operator perkalian (`*`).
* Jika promo valid, program menghitung potongan diskon menggunakan pengurangan (`-`).


4. **Output Akhir**: Program menampilkan rincian transaksi, total pembayaran akhir yang harus dibayar, serta status promo (apakah berhasil diterapkan atau gagal).

---

#### 7. Contoh Kondisi `True` dan `False`

`total_belanja = 150000` dengan kondisi pengecekan `total_belanja >= 100000` $\rightarrow$ Bernilai **`True`**.
