## 2.1 Dasar Teori

Struktur percabangan (Control Flow / Branching) adalah suatu instruksi dalam pemrograman yang memungkinkan alur eksekusi program berjalan secara dinamis berdasarkan suatu kondisi bernilai boolean (`True` atau `False`). Pada bahasa pemograman Python, terdapat beberapa jenis struktur percabangan yang digunakan untuk mengakomodasi berbagai tingkat kerumitan logika:

### 2.1.1 Pernyataan `if`

Digunakan untuk mengeksekusi satu blok kode tertentu hanya jika kondisi yang diuji bernilai `True`.

### 2.1.2 Pernyataan `if-else`

Menyediakan dua jalur eksekusi, di mana blok kode pertama dijalankan jika kondisi bernilai `True`, dan blok kode alternatif (`else`) dijalankan jika kondisi bernilai `False`.

### 2.1.3 Pernyataan `if-elif-else`

Digunakan ketika terdapat lebih dari dua alternatif kondisi yang harus diperiksa secara berurutan. Python akan mengevaluasi kondisi dari atas ke bawah dan berhenti pada kondisi pertama yang bernilai `True`.

### 2.1.4 Multi-if (If Berganda)

Merupakan sekumpulan pernyataan `if` terpisah yang dijalankan secara berurutan, sehingga setiap kondisi akan diuji terlepas dari hasil kondisi sebelumnya (memungkinkan beberapa blok kode tereksekusi sekaligus jika memenuhi syarat).

### 2.1.5 Pernyataan `match-case`

Diperkenalkan mulai Python 3.10, struktur ini berfungsi untuk mencocokkan nilai variabel terhadap sejumlah pola (`case`) secara lebih bersih dan efisien dibandingkan `if-elif-else` bertingkat.

### 2.2.4 Program match_case "A_2611531019"

```python
# Buat file dengan nama match_case1.py
# Buat program untuk match case
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# program konversi angka menjadi nama bulan

bulan = int(input("Masukkan angka bulan (1 - 12): "))

match bulan:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

### **Pejelasan**

1. **Baris 1 sampai 5** (`# ...`): Kumpulan baris komentar yang berisi instruksi pembuatan file bernama `match_case1.py`, catatan bahwa program menggunakan struktur `match case`, aturan penambahan 4 digit NIM terakhir pada penamaan variabel, keterangan penggunaan fungsi `input()`, serta penjelasan bahwa program ini berfungsi untuk mengonversi angka menjadi nama bulan.

3. **Baris 9** (`match bulan:`): Memulai struktur percabangan `match-case` untuk mengevaluasi nilai yang tersimpan di dalam variabel `bulan`.
4. **Baris 10 sampai 31** (`case 1: hingga case 12:`): Rangkaian pola pencocokan nilai (`case`) dari angka 1 sampai 12. Jika nilai variabel `bulan` cocok dengan salah satu angka tersebut, maka program akan mengeksekusi perintah `print()` di bawahnya untuk mencetak nama bulan yang sesuai (dari Januari hingga Desember).

### 2.2.6 Program multi_if2 "A_2611531019"

```python
# Buat file dengan nama multi_if2_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: total_belanja_1234
# Program ini menggunakan fungsi input()
# Program Menghitung Diskon Belanja

# Input dari user
total_belanja = float(input("Masukkan total belanja (Rp): "))

# Input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member = input("Apakah Anda member? (y/t): ").strip().lower()
is_member = input_member in ["y", "ya"]

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid = input_promo in ["y", "ya"]

total_diskon_persen = 0

# Multi-IF terpisah: Setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja > 1000000:
    total_diskon_persen += 10 # Diskon belanja besar

if is_member:
    total_diskon_persen += 5 #Diskon member

if kode_promo_valid:
    total_diskon_persen += 15 # Diskon voucher

#Menghitung nominal diskon dan total bayar
nominal_diskon = total_belanja * (total_diskon_persen / 100)
total_bayar = total_belanja - nominal_diskon

#Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon : {total_diskon_persen}% (Rp {nominal_diskon:,.0f})")
print(f"Total Bayar  : Rp {total_bayar:,.0f}")

print (f"Total diskon yang Anda dapatkan: {total_diskon_persen}%")
#Output: Total diskon yang Anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

### **Penjelasan :**

2. **Baris 7** (`total_belanja = float(input("Masukkan total belanja (Rp): "))`): Meminta pengguna memasukkan total belanja, lalu fungsi `float()` mengubah teks masukan tersebut menjadi tipe data desimal (float) dan menyimpannya ke dalam variabel `total_belanja`.
3. **Baris 10** (`input_member = input("Apakah Anda member? (y/t): ").strip().lower()`): Meminta input status member, di mana `.strip()` menghapus spasi berlebih dan `.lower()` mengubah huruf menjadi huruf kecil untuk memudahkan validasi.
4. **Baris 11** (`is_member = input_member in ["y", "ya"]`): Mengecek apakah nilai input member ada di dalam daftar [`"y", "ya"`], menghasilkan nilai boolean (`True` atau `False`) yang disimpan ke variabel `is_member`.

### 2.2.7 Program nilai "A_2611531019"

```python
# Buat file dengan nama nilai_nim.py
# Buat program untuk nested if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# program konversi nilai angka menjadi huruf

nilai_1019 = int(input("Inputkan nilai angka= "))

if nilai_1019 >= 81:
    print("A")
elif nilai_1019 >= 70:
    print("B")
elif nilai_1019 >= 60:
    print("C")
elif nilai_1019 >= 50:
    print("D")
else:
    print("E")
```

3. **Baris 9** (`if nilai_1019 >= 81:`): Memulai struktur percabangan bertingkat dengan memeriksa kondisi pertama, apakah nilai angka lebih besar dari atau sama dengan 81.
4. **Baris 10** (`print("A")`): Blok kode yang akan dieksekusi apabila nilai angka memenuhi syarat lebih besar dari atau sama dengan 81, yaitu mencetak huruf `"A"`.
5. **Baris 11** (`elif nilai_1019 >= 70:`): Cabang kondisi alternatif pertama (`else if`), memeriksa apakah nilai angka lebih besar dari atau sama dengan 70.
6. **Baris 12** (`print("B")`): Blok kode yang dieksekusi jika kondisi pada baris 11 terpenuhi, yaitu mencetak huruf `"B"`.
