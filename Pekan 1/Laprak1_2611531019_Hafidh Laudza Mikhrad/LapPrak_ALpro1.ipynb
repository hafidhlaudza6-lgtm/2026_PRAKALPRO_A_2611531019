## 1.1 Latar Belakang
Latar Belakang

Perkembangan teknologi informasi yang semakin pesat membuat kemampuan dalam bidang pemrograman menjadi salah satu keterampilan yang penting bagi mahasiswa IT. Pemrograman digunakan untuk membuat berbagai aplikasi dan sistem yang dapat membantu menyelesaikan berbagai kebutuhan manusia. Salah satu bahasa pemrograman yang banyak digunakan dan mudah dipelajari adalah Python karena memiliki sintaks yang sederhana serta dapat digunakan dalam berbagai bidang teknologi.

Dalam proses pemrograman, diperlukan aplikasi yang dapat membantu programmer dalam menulis dan menjalankan kode. Salah satu aplikasi yang banyak digunakan adalah Visual Studio Code (VS Code). VS Code menyediakan berbagai fitur yang memudahkan proses pengembangan program, seperti code editor, terminal, debugging, dan berbagai extension.

Selain kemampuan membuat program, mahasiswa IT juga perlu memahami cara mengelola dan menyimpan kode program. Git dan GitHub merupakan tools yang banyak digunakan untuk melakukan version control dan menyimpan project secara online. Dengan Git dan GitHub, perubahan pada kode dapat dicatat dan project dapat dikelola dengan lebih terstruktur.

Oleh karena itu, praktikum pengenalan Python, VS Code, dan koneksi ke GitHub dilakukan untuk memberikan pemahaman dasar mengenai proses pembuatan program sekaligus pengelolaan project menggunakan version control. Melalui praktikum ini, diharapkan mahasiswa dapat memiliki dasar keterampilan yang dapat digunakan dalam pengembangan project pemrograman selanjutnya.
