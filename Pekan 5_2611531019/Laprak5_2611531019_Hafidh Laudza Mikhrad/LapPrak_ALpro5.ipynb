## 2.1 Dasar Teori

Struktur kontrol perulangan (looping) merupakan salah satu pilar fundamental dalam pemrograman untuk mengeksekusi blok kode secara berulang-ulang berdasarkan kondisi atau batasan tertentu. Dalam bahasa pemrograman Python, terdapat dua mekanisme perulangan utama, yaitu for loop dan struktur perulangan bersarang (nested loop).

### 2.1.1 Perulangan `for` (`for loop`)

Perulangan `for` di Python digunakan untuk melakukan iterasi terhadap suatu urutan (sequence), seperti objek `range()`, `list`, `tuple`, atau `string`. Khusus untuk pengulangan berbasis angka terstruktur, fungsi `range(start, stop, step)` sangat diandalkan. Perulangan ini sangat efisien ketika jumlah iterasi sudah diketahui sebelumnya (counted loop).

### 2.1.2 Perulangan Bersarang (Nested Loop)

Nested loop adalah kondisi di mana sebuah perulangan (`for` atau `while`) berada di dalam badan perulangan lainnya. Perulangan luar (outer loop) dan perulangan dalam (inner loop) bekerja secara simultan, di mana untuk setiap satu iterasi perulangan luar, perulangan dalam akan menyelesaikan seluruh siklus iterasinya terlebih dahulu.

## 2.2 Kode Program

### 2.2.1 Program nested_for1 "A_2611531019"

```python
# Buat file dengan nama nested_for1_nim.py
# Buat program untuk perulangan for dalam pyton
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_1019 = int(input("Masukkan nilai batas: "))
for line in range(1, batas_1019 + 1):
    for j_1019 in range(1, (-1 * line + batas_1019) + 1):
        print(".", end="")
    print(line)
```

### **Penjelasan :**

1. Baris 5 (`batas_1019 = int(input("Masukkan nilai batas: "))`): Menerima masukan angka dari pengguna melalui konsol, lalu mengkonversinya ke dalam tipe data integer (bilangan bulat) dan menyimpannya ke dalam variabel `batas_1019`.
2. Baris 6 (`for line in range(1, batas_1019 + 1):`): Memulai perulangan luar (outer loop) dengan variabel kontrol `line` yang melakukan iterasi mulai dari angka 1 sampai dengan nilai `batas_1019`.
3. Baris 7 (`for j_1019 in range(1, (-1 * line + batas_1019) + 1):`): Memulai perulangan dalam (inner loop) bersarang dengan variabel `j_1019` yang menentukan jumlah karakter titik yang akan dicetak pada baris tersebut secara dinamis.
4. Baris 8 (`print(".", end="")`): Mencetak karakter titik (.) ke layar tanpa berpindah ke baris baru karena menggunakan parameter `end=""`.
5. Baris 9 (`print(line)`): Mencetak nilai dari variabel `line` ke konsol sekaligus berfungsi untuk memindahkan posisi kursor ke baris baru (newline) setelah perulangan titik selesai dicetak pada baris tersebut.

### 2.2.4 Program nasted_for4 "A_2611531019"

```python
# Buat file dengan nama nested_for4_nim.py
# Buat program untuk perulangan for dalam pyton
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_1019 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_1019 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_1019 = tinggi_1019
    c_1019 = a_1019
    lebar_1019 = (2 * tinggi_1019) - 2

    for i_1019 in range(1, tinggi_1019 + 1):
        b_1019 = c_1019 + 1

        for j_1019 in range(1, lebar_1019 + 1):

            # Baris atas dan bawah
            if i_1019 == 1 or i_1019 == tinggi_1019:
                if j_1019 == 1 or j_1019 == lebar_1019:
                    print("#", end="")
                else:
                    print("=", end="")

            # Baris isi
            else:
                if j_1019 == 1 or j_1019 == lebar_1019:
                    print("|", end="")
                else:
                    if j_1019 == c_1019:
                        print("<", end="")
                    elif j_1019 == b_1019:
                        print(">", end="")
                    elif j_1019 == (lebar_1019 - c_1019):
                        print("<", end="")
                    elif j_1019 == (lebar_1019 - c_1019 + 1):
                        print(">", end="")
                    elif j_1019 > b_1019 and j_1019 < (lebar_1019 - c_1019):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # Logika asli java
        a_1019 -= 2

        if a_1019 <= 0:
            c_1019 = (-a_1019) + 2
        else:
            c_1019 = a_1019
```

### **Pejelasan**

6. Baris 10 (`if i_1019 < ulang_1019:`): Melakukan percabangan untuk mengecek apakah iterasi belum mencapai angka terakhir (lebih kecil dari `ulang_1019`).

2. Penerapan Kompleksitas Data dengan Nested Loop:
Perulangan bersarang (nested loop), yaitu kondisi di mana satu atau lebih perulangan berada di dalam blok perulangan utama, sangat esensial untuk menyelesaikan pemrosesan data yang bersifat multidimensional atau kompleks. Mekanisme ini bekerja di mana perulangan bagian dalam akan diselesaikan terlebih dahulu secara penuh untuk setiap satu iterasi dari perulangan bagian luar, sehingga sangat ideal digunakan dalam pembuatan matriks, tabel, maupun pola karakter bertingkat.
3. Peningkatan Kemampuan Logika Pemrograman:
Praktikum ini berhasil membimbing praktikan dalam memahami alur kontrol logika iterasi secara mendalam. Pemahaman yang baik mengenai kombinasi perulangan `for` dan nested loop menjadi landasan krusial dalam merancang algoritma pemecahan masalah (problem solving) yang optimal dalam pengembangan perangkat lunak berbasis bahasa pemrograman Python.
